```python
import sympy as sp
import numpy as np 
import matplotlib.pyplot as plt

from IPython.display import display, Math
```

```python
x = sp.symbols('x')
print(x)
display(x)

alpha , beta = sp.symbols('alpha, beta')
print(alpha, beta)
display(alpha, beta)
```

```python
expr = alpha * x + beta * x **2
expr_latex = sp.latex(expr, mode ="equation")
display(expr)
sp.pprint(expr)
print(expr_latex)

display(Math(expr_latex))
```

```python
display(sp.pi, sp.E, sp.I, sp.oo) 

expr = sp.sin(x) + sp.exp(x **2) + sp.log(sp.sqrt(x))
display(expr)
```

```python
expr = x + sp.Integer(1)/3
display(expr)

expr = x + sp.Rational(1,3)
display(expr)
```

```python
x= sp.symbols('x', real= True, nonnegative= True)
expr = sp.sqrt(x ** 2)
display(expr)

y= sp.symbols('y')
expr = sp.sqrt(y ** 2)
display(expr)
```

```python
theta = sp.symbols('theta')
expr = sp.sin(theta)**2 + sp.cos(theta)**2
display(expr)
display(expr.simplify())
```

```python
x =sp.symbols('x')

poly= (x**2 - 1)**2 * (x**2 + 1) 

roots_poly = sp.solve(poly, x)
display(roots_poly)

roots2= sp.solveset(poly,x, domain =sp.S.Reals)
display(roots2)

roots3 = sp.roots(poly, x)
display(roots3) 
```

```python
x,y = sp.symbols('x y')
poly= (x**2 - 1)**2 * (y**2 + 1) 

diffx = sp.diff(poly, x)
display(diffx)
diffxy = sp.diff(poly, x, y)
display(diffxy)

diffx2y = sp.diff(poly, x,x, y)
display(diffx2y)

diffx2y_b = sp.diff(poly, x,2, y)
display(diffx2y_b)
```

```python
x,y = sp.symbols('x y')
poly= (x**2 - 1)**2 * (y**2 + 1) 

limx0 = sp.limit(poly, x, 0)
display(limx0)

limx0l = sp.limit(poly, x, 0, dir='-')
display(limx0)

limx0both = sp.limit(poly, x, 0, dir='+-')
display(limx0both)
```

```python
x= sp.symbols('x')
poly = (x**2 - 1)**2

poly_np = sp.lambdify(x, poly, 'numpy')

x_np = np.linspace(-2, 2, 100)
y_np = poly_np(x_np)

fig,ax = plt.subplots(figsize= (3,3))
ax.plot(x_np, y_np)
plt.show()

```
