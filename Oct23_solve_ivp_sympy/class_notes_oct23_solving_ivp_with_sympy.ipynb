```python
ode = sp.Eq(y(t).diff(t) + y(t) , -2)
sol = sp.dsolve(ode, ics={ y(0) :  0} )
display(sol)
print(sp.limit(sol.rhs, t, sp.oo))
sp.plot(sol.rhs, (t,0,10), size=(3,2), ylabel="y(t)")
```

```python
ode = sp.Eq(y(t).diff(t,2) + y(t).diff(t) + 2 * y(t) , sp.Rational(1,2))
sol = sp.dsolve(ode, ics ={y(0) : 1 , y(t).diff(t).subs(t,0) : 0 })
display(sol)
print(sp.checkodesol(ode, sol)) # (True,0)

ynp= sp.lambdify(t, sol.rhs, "numpy")
dynp =sp.lambdify(t, sol.rhs.diff(t), "numpy")
tt = np.linspace(0,20,200)

fig,axs = plt.subplots(1,2, figsize=(10,3))
axs[0].plot(tt, ynp(tt), label="y(t)")
axs[0].set_xlabel("t")
axs[0].set_ylabel("y(t)") 

axs[1].plot(ynp(tt), dynp(tt))
axs[1].set_xlabel("y(t)")
axs[1].set_ylabel("y'(t)")
axs[1].set_title("phase portrait")

axs[0].legend()
axs[1].legend()

plt.show()
```
