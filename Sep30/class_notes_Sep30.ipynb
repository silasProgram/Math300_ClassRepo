```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

Polynomial complexity growth $O(n^a)$ with $n$ being teh size of the input
\begin{align}
 t &= c \, n^a ,\\
 \log(t) &= a \, \log(n) + \log(c)
\end{align}

plot `loglog(n, t)`

` slope, intercept = np.polyfit(log(n), log(t), 1)`

`c = np.exp(intercept)`

```python
def t_poly(n, a, c):
    return  c* n**a

n = np.arange(1, 101)
a = 2
c = 1
y = t_poly(n, a, c)
slope, intercept = np.polyfit(np.log(n), np.log(y), 1)

fig, axs = plt.subplots(figsize=(4,4))
axs.loglog(n, y, "-o", label="poly growth")
axs.loglog(n, np.exp(intercept)* n**slope, "--", label="Fit")
axs.legend()

axs.set_xlabel("n")
plt.show()
```

Exponential complexity growth $O(b^n)$ with $b>1$
\begin{align}
 t &= c \, b^n ,\\
 \log(t) &= n \, \log(b) + \log(c)
\end{align}

plot `semilogy(n, t)`

` slope, intercept = np.polyfit(n, log(t), 1)`

`c = np.exp(intercept)`

`b= np.exp(slope)`

```python
def t_exp(n, b, c):
    return  c* b**n

n = np.arange(1, 20)
b =2
c = 1
y = t_exp(n, b, c)

slope, intercept = np.polyfit(n, np.log(y), 1)

fig, axs = plt.subplots(figsize=(4,4))
axs.semilogy(n, y, "-o", label="exp growth")
axs.semilogy(n, np.exp(intercept)* np.exp(slope)**n, "--", label="Fit")
axs.legend()
axs.set_xlabel("n")
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(8,4))

axs[0].loglog(n, t_poly(n, a, c), "--", label="poly growth")
axs[0].loglog(n, t_exp(n, b, c), ":", label="exp growth")

axs[1].semilogy(n, t_poly(n, a, c), "--", label="poly growth")
axs[1].semilogy(n, t_exp(n, b, c), ":", label="exp growth")


axs[0].legend()
axs[1].legend()
plt.show()
```

$|x_{k} - x_{*}| < \tau $ for large k 


$[x_0, x_1, x_2 , \ldots ]$

$x_{k+1} = \frac{1}{2} ( x_{k} + \frac{2}{x_k}) $  

$g(x_{*}) =x_{*}$



$x_{*} = \frac{1}{2} ( x_{*} + \frac{2}{x_{*} }) $ 

```python
def seq_rule(x):
    return (x + 2./x)/2.

x0 =3.0
x =[x0]
err = [np.abs(x[-1] - np.sqrt(2))]
errdiff = []
for idx in range(12):
    x.append(seq_rule(x[-1]))
    err.append(np.abs(x[-1] - np.sqrt(2)))
    errdiff.append(np.abs(x[-1] - x[-2]))

print(x[-4:], np.sqrt(2))

fig, ax = plt.subplots(figsize=(4,4))
ax.semilogy(err, label="error")
ax.semilogy(errdiff, label="error difference")
ax.legend()
plt.show()
```
