question 3: secret number is 68

```python
p = np.random.default_rng(seed=0)
normal_sample_mean_5 = p.normal(loc=5, scale =2, size = 200)
thresh = 5
#print(normal_sample_mean_5)
def count_above(arr, threshold):
    exceed = 0
    for i in arr:
        if i > threshold:
            exceed += 1
    return exceed
test = count_above(normal_sample_mean_5, thresh)
x = np.sum(normal_sample_mean_5 > thresh)
print(f"my value is found to be {test} and the vectorized approach returns {x}")
```
