```python
import numpy as np
import scipy.interpolate as si
import matplotlib.pyplot as plt
```

```python
ndate = 7
x = np.linspace(-4,4, ndata)
y = np.arctan(x)

fig,ax = plt.subplots(figsize=(4,4))
ax.plot(x,y, 'ko', label="data")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
xnew = np.linspace(-4,4, 101)
plin = lambda xin: np.interp(xnew, x, y)

fig,ax = plt.subplots(figsize=(4,4))
ax.plot(x,y, 'ko', label="data")
ax.plot(xnew, plin(xnew), label="linear interpolation")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
pcubic = si.CubicSpline(x,y)

fig,ax = plt.subplots(figsize=(4,4))
ax.plot(x,y, 'ko', label="data")
ax.plot(xnew, pcubic(xnew), label="cubic spline")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
ppchip = si.PchipInterpolator(x,y)

fig,ax = plt.subplots(figsize=(4,4))
ax.plot(x,y, 'ko', label="data")
ax.plot(xnew, ppchip(xnew), label="PCHIP")
ax.plot(xnew, np.arctan(xnew),"--" ,label="true function")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
pakima = si.Akima1DInterpolator(x,y, method="akima")
pmakima = si.Akima1DInterpolator(x,y, method="makima")

fig,ax = plt.subplots(figsize=(4,4))
ax.plot(x,y, 'ko', label="data")
ax.plot(xnew, pakima(xnew), label="akima")
ax.plot(xnew, pmakima(xnew), label="makima")
ax.plot(xnew, np.arctan(xnew),"--" ,label="true function")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```
